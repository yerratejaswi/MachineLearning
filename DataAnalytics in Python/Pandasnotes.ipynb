```python
import pandas as pd
# creating a series: series stores the data irrespective of its data type:
data={'a':1,'b':'Tejaswi','c':3,'d':4}
series_dict=pd.Series(data)
print(series_dict)
```

```python
#DataFrame:creating a df from dictionary:
data={
    'Name':['Tejaswi','Dinesh','Vinayak'],
    'Age': [25,30,45],
    'City':['NewYork','Bangalore','Texas']
}

data1=pd.DataFrame(data)
print(data1)
print(type(data1))
```

```python
#creating a df from list of dictinaries;

data=[
    {'Name':'Tejaswi','Age':55,'City':'Hyderbad'},
    {'Name':'Dinesh','Age':30,'City':'Chicago'},
    {'Name':'Vinayak','Age':21,'City':'Mumbai'},
    {'Name':'Siddartha','Age':22,'City':'Hyderbad'}
]

df = pd.DataFrame(data)
print(df)
```
